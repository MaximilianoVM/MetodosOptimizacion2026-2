```python
from sudoku import sudoku
import ipywidgets as widgets
from IPython.display import display, HTML

# ---------- Cuadrícula de entrada ----------
def crear_celda(r, c):
    derecha = '12px' if c in (2, 5) else '2px'
    abajo   = '12px' if r in (2, 5) else '2px'
    celda = widgets.Text(
        placeholder='·',
        layout=widgets.Layout(width='35px', margin=f'0 {derecha} {abajo} 0')
    )
    celda.add_class(f'sdk-{r}-{c}')   # etiqueta con su posición, para las flechas
    return celda

celdas = [[crear_celda(r, c) for c in range(9)] for r in range(9)]

boton = widgets.Button(description='Resolver')
salida = widgets.Output()

def a_cuadrantes(T):
    """Convierte un tablero 9x9 en tu lista de 9 bloques 3x3 (A, B, ..., I)."""
    return [[row[3*q:3*q+3] for row in T[3*p:3*p+3]]
            for p in range(3) for q in range(3)]

# ---------- Mostrar la solución con forma de sudoku ----------
def formatear(S, T):
    """S: solución 9x9. T: tablero original (para marcar las pistas con *)."""
    linea = '+---------+---------+---------+'
    texto = [linea]
    for r in range(9):
        fila = '|'
        for c in range(9):
            marca = '*' if T[r][c] != 0 else ' '
            fila += f' {S[r][c]}{marca}'
            if c in (2, 5, 8):
                fila += '|'
        texto.append(fila)
        if r in (2, 5, 8):
            texto.append(linea)
    return '\n'.join(texto)

def al_presionar(_):
    with salida:
        salida.clear_output()
        T = [[int(c.value) if c.value.strip().isdigit() else 0 for c in fila]
             for fila in celdas]
        S = sudoku(a_cuadrantes(T))
        if S is None:
            print('Infactible: revisa los números')
        else:
            print(formatear(S, T))
            print('* = número dado')

boton.on_click(al_presionar)

display(widgets.VBox([widgets.HBox(fila) for fila in celdas] + [boton, salida]))
```
