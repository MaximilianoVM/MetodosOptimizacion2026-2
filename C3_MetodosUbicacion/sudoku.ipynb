# Sudoku 

<img src="sudoku_real.jpeg" width="400">

$$ \begin{bmatrix} A & B & C \\ D & E & F \\ G & H & I \end{bmatrix} $$

donde 

$$A = [a_{ijk}] = \begin{bmatrix} a_{11k} & a_{12k} & a_{13k} \\ a_{21k} & a_{22k} & a_{23k} \\ a_{31k} & a_{32k} & a_{33k} \end{bmatrix}$$

donde 

$$a_{ijk} = 
\begin{cases}
	1 & \text{si la posicion $(i,j)$ tiene el numero $k$}
	\\
	0 & \text{si no}
\end{cases}$$

y 

$$i \in \{1, 2, 3\}$$
$$j \in \{1, 2, 3\}$$
$$k \in \{1, 2, 3, 4, 5, 6, 7, 8, 9\}$$

```python
from docplex.mp.model import Model
import numpy as np

mdl = Model(name='Sudoku')

import numpy as np
```

```python
A = [[0, 0, 0],
      [4, 3, 0],
      [8, 0, 0]]

B = [[0, 1, 0],
      [0, 6, 0],
      [3, 0, 4]]

C = [[0, 9, 0],
      [5, 0, 0],
      [0, 1, 6]]

D = [[0, 0, 4],
      [0, 0, 5],
      [1, 0, 0]]

E = [[0, 8, 0],
      [0, 2, 0],
      [6, 0, 0]]

F = [[0, 0, 9],
      [1, 0, 0],
      [0, 0, 0]]

G = [[0, 0, 8],
      [2, 0, 0],
      [7, 0, 0]]

H = [[0, 0, 9],
      [0, 0, 5],
      [0, 0, 0]]

I = [[0, 0, 0],
      [0, 0, 0],
      [0, 8, 0]]
```

```python
sudoku = np.block([[A, B, C],
                   [D, E, F],
                   [G, H, I]])

sudoku[1,1]
```

```python
I_i = range(3)        # renglón dentro del bloque
J_j = range(3)        # columna dentro del bloque
K_k = range(1, 10)    # número 1..9
```

```python
a = mdl.binary_var_cube(I_i, J_j, K_k, name='a')
b = mdl.binary_var_cube(I_i, J_j, K_k, name='b')
c = mdl.binary_var_cube(I_i, J_j, K_k, name='c')

d = mdl.binary_var_cube(I_i, J_j, K_k, name='d')
e = mdl.binary_var_cube(I_i, J_j, K_k, name='e')
f = mdl.binary_var_cube(I_i, J_j, K_k, name='f')

g = mdl.binary_var_cube(I_i, J_j, K_k, name='g')
h = mdl.binary_var_cube(I_i, J_j, K_k, name='h')
i = mdl.binary_var_cube(I_i, J_j, K_k, name='i')
```

### Restricciones: filas 

Un numero determinado $k$ no se debe repetir en una misma fila determinada $i$

$$\sum_j (a_{ijk} + b_{ijk} + c_{ijk}) = 1 \quad \forall i, k$$

$$\sum_j (d_{ijk} + e_{ijk} + f_{ijk}) = 1 \quad \forall i, k$$

$$\sum_j (g_{ijk} + h_{ijk} + i_{ijk}) = 1 \quad \forall i, k$$

```python
for m in I_i:
    for k in K_k:
        mdl.add_constraint(mdl.sum(a[m, j, k] + b[m, j, k] + c[m, j, k] for j in J_j) == 1)

for m in I_i:
    for k in K_k:
        mdl.add_constraint(mdl.sum(d[m, j, k] + e[m, j, k] + f[m, j, k] for j in J_j) == 1)
        
for m in I_i:
    for k in K_k:
        mdl.add_constraint(mdl.sum(g[m, j, k] + h[m, j, k] + i[m, j, k] for j in J_j) == 1)
```

### Restricciones: Columnas 

Un numero determinado $k$ no se debe repetir en una misma columna determinada $j$


$$\sum_i (a_{ijk} + d_{ijk} + g_{ijk}) = 1 \quad \forall j, k$$

$$\sum_i (b_{ijk} + e_{ijk} + h_{ijk}) = 1 \quad \forall j, k$$

$$\sum_i (c_{ijk} + f_{ijk} + i_{ijk}) = 1 \quad \forall j, k$$

```python
for j in J_j:
    for k in K_k:
        mdl.add_constraint(mdl.sum(a[m, j, k] + d[m, j, k] + g[m, j, k] for m in I_i) == 1)

for j in J_j:
    for k in K_k:
        mdl.add_constraint(mdl.sum(b[m, j, k] + e[m, j, k] + h[m, j, k] for m in I_i) == 1)
        
for j in J_j:
    for k in K_k:
        mdl.add_constraint(mdl.sum(c[m, j, k] + f[m, j, k] + i[m, j, k] for m in I_i) == 1)
```

### Restricciones: cuadrito

$$\sum_i \sum_j a_{ijk} = 1 \quad \forall k \quad ; \quad \sum_i \sum_j b_{ijk} = 1 \quad \forall k$$
$$\sum_i \sum_j c_{ijk} = 1 \quad \forall k \quad ; \quad \sum_i \sum_j d_{ijk} = 1 \quad \forall k$$

$$\sum_i \sum_j e_{ijk} = 1 \quad \forall k$$

$$\sum_i \sum_j f_{ijk} = 1 \quad \forall k \quad ; \quad \sum_i \sum_j g_{ijk} = 1 \quad \forall k$$
$$\sum_i \sum_j h_{ijk} = 1 \quad \forall k \quad ; \quad \sum_i \sum_j i_{ijk} = 1 \quad \forall k$$

### Restricciones: cuadricula 
Cada cuadricula $i,j$ solo puede tomar un valor $k$ 

$$\sum_k a_{ijk} = 1 \quad \forall i,j \quad ; \quad \sum_k b_{ijk} = 1 \quad \forall i,j \quad ; \quad \sum_k c_{ijk} = 1 \quad \forall i,j$$
$$\sum_k d_{ijk} = 1 \quad \forall i,j \quad ; \quad \sum_k e_{ijk} = 1 \quad \forall i,j \quad ; \quad \sum_k f_{ijk} = 1 \quad \forall i,j$$
$$\sum_k g_{ijk} = 1 \quad \forall i,j \quad ; \quad \sum_k h_{ijk} = 1 \quad \forall i,j \quad ; \quad \sum_k i_{ijk} = 1 \quad \forall i,j$$

```python
bloques = [a, b, c, d, e, f, g, h, i]

for p in bloques: 
    for m in I_i:
        for j in J_j:
            mdl.add_constraint(mdl.sum(p[m,j,k] for k in K_k) == 1)
```

```python
def leer_bloque(var):
    """Convierte las variables de un bloque en una matriz 3x3 con los números."""
    M = np.zeros((3, 3), dtype=int)
    for m in I_i:
        for j in J_j:
            for k in K_k:
                if var[m, j, k].solution_value > 0.5:
                    M[m, j] = k
    return M

R = [leer_bloque(p) for p in bloques]

solucion = np.block([[R[0], R[1], R[2]],
                     [R[3], R[4], R[5]],
                     [R[6], R[7], R[8]]])
print(solucion)
```
