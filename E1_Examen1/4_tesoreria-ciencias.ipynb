# Tesoreria de la olimpiada de ciencias

```python
from docplex.mp.model import Model
import numpy as np

mdl = Model(name='Tesoreria de la olimpiada de ciencias')
```

## Variables de decision


### Inversiones
* $x_{ij}$ <- $ invertido en cada instrumento $i$ en el mes $j$

donde 

$i \epsilon \{A, B, C\}$

$j \epsilon \{1, 2, 3, 4, 5, 6, 7\}$

```python
instrumento = ['A', 'B', 'C']
mes = range(1, 7+1)
```

```python
x = mdl.continuous_var_dict(
    [(i, t) for i in instrumento for t in mes],
    lb=0, # positivos
    name='x'
)
```

```python
x
```

### creditos

* $c_j$ <- $ en credito prestado pedido en el mes $j$


```python
c = mdl.continuous_var_dict(
    [t for t in mes],
    lb=0, # positivos
    name='c'
)
```

```python
c
```

## Restricciones

### conservacion de flujo por meses

```python
# mes 1
mdl.add_constraint(
    c[1] + 380000 == x['A', 1] + x['B', 1] + x['C', 1] + 10000
)

# mes 2
mdl.add_constraint(
    c[2] + 40000 + 1.008*x['A', 1] == x['A', 2] + x['B', 2] + x['C', 2] + 1.015*c[1] + 10000
)

# mes 3 (C no se ofrece en mes 3)
mdl.add_constraint(
    c[3] + 0 + 1.008*x['A', 2] + 1.02*x['B', 1] == x['A', 3] + x['B', 3] + 1.015*c[2] + 120000 + 10000
)

# mes 4
mdl.add_constraint(
    c[4] + 60000 + 1.008*x['A', 3] + 1.02*x['B', 2] + 1.036*x['C', 1] == x['A', 4] + x['B', 4] + x['C', 4] + 1.015*c[3] + 90000 + 10000
)   

# mes 5 (C no se ofrece en mes 5)
mdl.add_constraint(
    c[5] + 0 + 1.008*x['A', 4] + 1.02*x['B', 3] + 1.036*x['C', 2] == x['A', 5] + x['B', 5] + 1.015*c[4] + 150000 + 10000
)   

# mes 6 (Agregamos 1.02*x['B', 4] que vence este mes; C no se ofrece y B no se contrata aquí)
mdl.add_constraint(
    c[6] + 20000 + 1.008*x['A', 5] + 1.02*x['B', 4] == x['A', 6] + 1.015*c[5] + 110000 + 10000
)   

# mes 7
#mdl.add_constraint(
#    1.008*x['A', 6] + 1.02*x['B', 5] + 1.036*x['C', 4] == 1.015*c[6] + 10000
#)  

mdl.add_constraint(
    1.008*x['A', 6] + 1.02*x['B', 5] + 1.036*x['C', 4] >= 1.015*c[6] + 10000
)   
```

### limite de credito por  mes

```python
# limite de credito por mes  
mdl.add_constraints(
    c[t] <= 50000 for t in mes
)
```

## Funcion objetivo

```python
mdl.maximize(
    1.008*x['A', 6] + 1.02*x['B', 5] + 1.036*x['C', 4] - 1.015*c[6] - 10000
)

# Resolver e imprimir solucion
solution = mdl.solve(log_output=True)


print(solution)
```
