# 1. Sustrato para cultivo de setas

```python
from docplex.mp.model import Model
import numpy as np

mdl = Model(name='Sustrato para cultivo de setas')
```

## Variables de decision
kg de ingrediente $i$ utilizado por semana

```python
ingredientes_i=[0, 1, 2, 3, 4, 5]

# Variable de decision
x = mdl.integer_var_list(len(ingredientes_i), name='x')  # por defecto, x_i >= 0 y sin límite superior
```

```python
x
```

## Restricciones

### i)
$$\sum_i x_i == 2000$$

```python
mdl.add_constraint(
    mdl.sum(x[i] for i in ingredientes_i) == 2000
    )
```

### ii)

```python
mdl.add_constraint(
    0.005*x[0] + 0.007*x[1] + 0.003*x[2] + 0.018*x[3] + 0.026*x[4] >= 0.008*2000
    )
```

```python
mdl.add_constraint(
    0.005*x[0] + 0.007*x[1] + 0.003*x[2] + 0.018*x[3] + 0.026*x[4] <= 0.014*2000
    )
```

### iii)

```python
mdl.add_constraint(
    0.45*x[0] + 0.44*x[1] + 0.46*x[2] + 0.42*x[3] + 0.40*x[4] >= 0.30*2000
)
```

```python
mdl.add_constraint(
    0.45*x[0] + 0.44*x[1] + 0.46*x[2] + 0.42*x[3] + 0.40*x[4] <= 0.60*2000
)
```

### iv)

```python
mdl.add_constraint(
    x[5] >= 0.01*2000
)

mdl.add_constraint(
    x[5] <= 0.02*2000
)
```

### v)

```python
mdl.add_constraint(
    x[3] <= 0.25*2000
)
```

### vi)

```python
mdl.add_constraint(
    x[4] <= 0.08*2000
)
```

### vii) Disponibilidad

```python
mdl.add_constraint(
    x[1] <= 900
)

mdl.add_constraint(
    x[2] <= 600
)

mdl.add_constraint(
    x[3] <= 400
)
```

## Funcion objetivo

```python
# Función Objetivo
mdl.minimize(
    3*x[0] + 2.2*x[1] + 1.5*x[2] + 6.5*x[4] + 4*x[5]
)
```

```python
solution = mdl.solve(log_output=True,)

print(solution)
```
