# 1. Asignación de órdenes a máquinas

La empresa de plásticos ha recibido cinco **órdenes** *(A1 , A2 , A3 , A4 , A5 )* que deben completarse.

Para atenderlas se dispone de cinco **máquinas** *(M1 , M2 , M3 , M4 , M5)*. 

Cada máquina puede realizar cada tarea con el **costo** que se indica en la *Tabla 1*. El problema consiste en determinar la asignación óptima que **minimice el costo total** de completar las órdenes, suponiendo que **cada máquina solo puede procesar una orden** y que **todas las órdenes deben ser atendidas.**


a) Formule un modelo de programación lineal entera que determine el plan de asignación óptimo.

b) Suponga que procesar cada orden en cada máquina implica un costo promedio de preparación de 10 unidades. ¿Cómo se modifica el modelo del inciso anterior?

```python
from docplex.mp.model import Model
import numpy as np

mdl = Model(name='Asignacion de ordenes a maquinas')

# Tabla 1
costos = [
    # M1  M2  M3   M4    M5
    [16,   4,  9,   5,   6],  # A1
    [2,   14,  7,   5,  13],  # A2
    [8,   10,  3,  12,  11],  # A3
    [3,    7,  6,  10,   5],  # A4
    [3,    6,  8,  11,   7],  # A5
]

costos_np = np.array(costos)

costos_np
```

## Datos y supuestos  

* cinco **órdenes** *(A1 , A2 , A3 , A4 , A5 )* 
* cinco **máquinas** *(M1 , M2 , M3 , M4 , M5)*
* Cada máquina puede realizar cada tarea con el **costo** que se indica en la *Tabla 1*
* determinar la asignación óptima que **minimice el costo total** de completar las órdenes
* **cada máquina solo puede procesar una orden**
* **todas las órdenes deben ser atendidas.**

## Formulación matemática 
### Variables de decisión
Variables binarias; La orden $i$ se atiende en que maquina $j$ ? 

$$x_{ij} =
\begin{cases}
	1 & \text{si se atiende la orden $i$ en la maquina} j
	\\
	0 & \text{si no}
\end{cases}$$
donde $$i\ \epsilon\ \{0, 1, 2, 3, 4\} = \{A_1, A_2, A_3, A_4, A_5\}$$ y 
$$j\ \epsilon\ \{0, 1, 2, 3, 4\} = \{M_1, M_2, M_3, M_4, M_5\}$$

```python
# ordenes y maquinas
num_ordenes = costos_np.shape[0]  # 5 filas
num_maquinas = costos_np.shape[1] # 5 columnas
```

```python
ordenes_i = range(num_ordenes)   # [0, 1, 2, 3, 4]
maquinas_j = range(num_maquinas) # [0, 1, 2, 3, 4]

x = mdl.binary_var_matrix(ordenes_i, maquinas_j, name='x')

x
```

### Restricciones
**Restriccion 1:**
* cada máquina solo puede procesar una orden
*  aquí $j$ está fijo (una maquina específica) y sumamos sobre todas las ordenes $i$.
$$\sum_{i} x_{ij} = 1$$

```python
# restricciones: cada maquina puede procesar a lo sumo una orden
mdl.add_constraints(
    mdl.sum(x[i,j] for i in ordenes_i) == 1 for j in maquinas_j
    )
```

**Restriccion 2:** 
* todas las órdenes deben ser atendidas.
* aquí $i$ está fijo (una orden específica) y sumamos sobre todas las máquinas $j$.
$$\sum_{j} x_{ij} = 1$$

```python
# restricciones: cada orden debe ser atendida
mdl.add_constraints(
    mdl.sum(x[i,j] for j in maquinas_j) == 1 for i in ordenes_i
    )
```

## Función objetivo 
* Cada máquina puede realizar cada tarea con el **costo** que se indica en la *Tabla 1*
* determinar la asignación óptima que **minimice el costo total** de completar las órdenes
### Para el caso a)

$$min\ z= \sum_{i} \sum_{j} c_{ij} x_{ij}​$$
### Para el caso b)
$$min\ z= \sum_{i} \sum_{j} (c_{ij}+10) x_{ij}​$$
$$= \sum_{i} \sum_{j} c_{ij}\ x_{ij} +  50$$
La asignación óptima no debería cambiar respecto al caso *a)* al solo sumar una constante. 

```python
mdl.minimize(
    mdl.sum(costos_np[i,j] * x[i,j] for i in ordenes_i for j in maquinas_j)
    )


# Resolver e imprimir solucion
solution = mdl.solve(log_output=True,)


print(solution)
```
