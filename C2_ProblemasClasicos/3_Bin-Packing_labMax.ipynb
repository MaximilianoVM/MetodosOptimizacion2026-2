# The bin-packing problem

El problema consiste en asignar cada objeto j en un contenedor, sujeto a la restriccion de no exceder la capacidad de los contenedores, y que el numero de contenedores utilizados sea el minimo.

## Datos y supuestos  
* conjunto de $n$ objetos, donde cada objeto $i$ tiene asociado un tamaño $w_j$ .
*  conjunto de $m$ contenedores $j$ (bins), con una capacidad de tamaño $W$.

Asignar cada objeto j en un contenedor, sujeto a la restricción de no exceder la capacidad de los contenedores, y que el número de contenedores utilizados sea mínimo.
## Formulación matemática 

```python
from docplex.mp.model import Model

mdl = Model(name='P empaquetamiento')

# pesos -> sacados de la pagina
pesos_w = [33, 38, 37, 31, 26, 12, 40, 12,
         19, 35, 14, 40, 11, 40, 30, 34,
         14, 27, 24, 31, 36, 25, 18, 29,
         14, 30, 24, 32, 34, 66]


# caso Homo:
capacidad_W = 100
capacidades_W = [capacidad_W for j in range(len(pesos_w))]
print('capacidades_W =', capacidades_W)

# indexamos contenedores
contenedores_j = [j for j in range(len(capacidades_W))] #Contenedores
print('contenedores_j =', contenedores_j)

# indexamos objetos
objetos_i = [i for i in range(len(pesos_w))] #Objetos
print('objetos_i =', objetos_i)
```

### Variables de decisión

$$x_{ij} = 
\begin{cases}
	1 & \text{si se coloca el objeto $i$ en el bin} j
	\\
	0 & \text{si no}
\end{cases}$$

```python
# Definir variables de decision
x = mdl.binary_var_matrix(objetos_i, contenedores_j, name='x')
x
```

```python
x[0,1]
```

$$y_{j} = 
\begin{cases}
	1 & \text{si el bin $j$ se usa ("activa")}
	\\
	0 & \text{si no}
\end{cases}$$

```python
# Definir variables de decision
y = mdl.binary_var_list(contenedores_j, name='y')
y
```

### Restricciones
#### Restricciones 1: Sobre $y_j$  - "Enlace" (Link Constraint)


Link Constraint
* Para que la variable $y_j$ funcione, debe existir una restricción que la obligue a "**encenderse**" ($y_j = 1$) en cuanto metes al menos un objeto en el contenedor $i$:
* Si no metes nada en el contenedor $j$ (todos los $x_{ij} = 0$), la suma de pesos es $0$. Como minimizamos $\sum y_j$, el solver elegirá $y_j = 0$.
* Si metes **aunque sea un solo objeto** en el contenedor $j$, la suma es $> 0$. Para que la desigualdad se cumpla, el solver se ve **obligado a encender $y_j = 1$** (multiplicando la capacidad total $W$ por 1).

$$\sum_{i=1}^{n} w_i x_{ij} \le W \cdot y_j \ \ \ \forall \ \ j$$



```python
# Definir las restricciones

mdl.add_constraints(mdl.sum(pesos_w[i]*x[i,j] for i in objetos_i) <= capacidad_W*y[j] for j in contenedores_j)
```

#### Restricciones 2
Un objeto $i$ se asigna solamente a un bin $j$
* Para cada objeto $i$ ( $\forall \ i$ ) verificamos que la suma, revisando todos los bins, sea 1.
* Esto asegura que cada objeto fue empacado, ademas de no estar duplicado.
$$\sum_j^m x_{ij} = 1 \ \ \forall \ i$$

```python
# Definir las restricciones

mdl.add_constraints(mdl.sum(x[i,j] for j in contenedores_j) == 1 for i in objetos_i)
```

## Función objetivo 
$$\min z= \sum_j^m y_j​$$

```python
# Definir la funcion objetivo
mdl.minimize(mdl.sum(y[j] for j in contenedores_j))
```

```python
solution = mdl.solve(log_output=True,)
print(solution)
```
