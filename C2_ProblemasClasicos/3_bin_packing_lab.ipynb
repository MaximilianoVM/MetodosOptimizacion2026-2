```python
# Import functions from the docplex module
from docplex.mp.model import Model
```

```python
mdl = Model(name='P empaquetamiento')
#conjuntos
pesos = [33, 38, 37, 31, 26, 12, 40, 12,
         19, 35, 14, 40, 11, 40, 30, 34,
         14, 27, 24, 31, 36, 25, 18, 29,
         14, 30, 24, 32, 34, 66]

obj = [(i) for i in range(len(pesos))] # objetos

obj
```

```python
# homo
#contenedores = [(j) for j in range(len(pesos))] #Contenedores
#contenedores
#capacidad = 100

# no homo
capacidad = [66, 100, 77, 69, 78, 93, 97, 89, 83, 64,
               82, 96, 78, 76, 75, 73, 61, 97, 99, 95,
               93, 96, 66, 70, 72, 96, 87, 75, 82, 89]
contenedores = [(j) for j in range(len(capacidad))] #Contenedores

x = mdl.binary_var_matrix(obj,contenedores,name='x')
y = mdl.binary_var_list(contenedores,name='y')
```

```python
# homo
#mdl.add_constraints(
#    mdl.sum(pesos[i]*x[i,j] for i in obj) <= (capacidad*y[j]) for j in contenedores
#    )

# no homo
mdl.add_constraints(
    mdl.sum(pesos[i]*x[i,j] for i in obj) <= (capacidad[j]*y[j]) for j in contenedores
    )

#mdl.add_constraints(
#    mdl.sum(x[i,j] for i in obj) == 1 for j in contenedores
#    )

mdl.add_constraints(
    mdl.sum(x[i,j] for j in contenedores) == 1 for i in obj
)

mdl.minimize(
    mdl.sum(y[j] for j in contenedores)
    )


# Resolver e imprimir solucion
solution = mdl.solve(log_output=True,)


print(solution)
```
